### Inspecting the DCT Coefficient Shapes

Before building the CNN model, we need to inspect the shapes of the extracted DCT coefficients for the Y, Cb, and Cr channels. These channels represent different components of the JPEG image: Y contains luminance information, while Cb and Cr contain chrominance information.

The coefficient arrays may have different spatial dimensions because JPEG images can use chroma subsampling. Checking their shapes helps us understand the data structure and determine how to prepare the inputs for the CNN.

This step is important because the model processes the Y channel separately from the two chrominance channels. It also helps us identify potential shape inconsistencies that could cause errors when loading samples into batches.

```python
import numpy as np
import os

root = "dct_dataset"

for split in ["train", "test"]:
    split_dir = os.path.join(root, split)

    for label in os.listdir(split_dir):
        label_dir = os.path.join(split_dir, label)

        if not os.path.isdir(label_dir):
            continue

        files = [
            f for f in os.listdir(label_dir)
            if f.endswith(".npy")
        ]

        if files:
            sample = np.load(
                os.path.join(label_dir, files[0]),
                allow_pickle=True
            ).item()

            print("Split:", split)
            print("Class:", label)
            print("Y shape:", sample["Y"].shape)
            print("Cb shape:", sample["Cb"].shape)
            print("Cr shape:", sample["Cr"].shape)
            break

    break
```
