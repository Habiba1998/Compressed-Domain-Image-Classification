```python
import os
from tensorflow.keras.datasets import cifar10
from PIL import Image

# Load CIFAR-10
(x_train, y_train), (x_test, y_test) = cifar10.load_data()

quality = 100

class_names = [
    "airplane",
    "automobile",
    "bird",
    "cat",
    "deer",
    "dog",
    "frog",
    "horse",
    "ship",
    "truck",
]

# Save training set
for i in range(len(x_train)):
    img = Image.fromarray(x_train[i])
    label = class_names[int(y_train[i][0])]

    save_dir = f"jpeg_dataset/train/{label}"
    os.makedirs(save_dir, exist_ok=True)

    img.save(
        f"{save_dir}/{i}.jpeg",
        format="JPEG",
        quality=quality
    )

# Save test set
for i in range(len(x_test)):
    img = Image.fromarray(x_test[i])
    label = class_names[int(y_test[i][0])]

    save_dir = f"jpeg_dataset/test/{label}"
    os.makedirs(save_dir, exist_ok=True)

    img.save(
        f"{save_dir}/{i}.jpeg",
        format="JPEG",
        quality=quality
    )
```
